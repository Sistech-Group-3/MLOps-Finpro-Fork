```python
import osmnx as ox
import networkx as nx
import joblib
```

```python
G = ox.graph_from_place("Chicago, Illinois, USA", network_type="walk")
```

```python
G = ox.add_edge_speeds(G)
G = ox.add_edge_travel_times(G)
G = nx.DiGraph(G)
```

```python
joblib.dump(G, "../graph/chicago_walk_graph.joblib", compress=3)
```

```python
G_loaded = joblib.load("../graph/chicago_walk_graph.joblib")
```
