>  due to a big amount of data records, I will sample a subset of the dataset

The key considerations for sampling:

1. 🔑 **Representative both feature wise and record wise**
	- why: to minimize the risk of information loss
	- how: stratified sampling

2. 💻 **Faster computation**
	- why: efficiency and manageability

```python
def compare_distributions(orig, sampled, name):
    o = orig.value_counts(normalize=True).sort_index()
    s = sampled.value_counts(normalize=True).sort_index()
    idx = o.index.union(s.index)
    o = o.reindex(idx, fill_value=0)
    s = s.reindex(idx, fill_value=0)
    max_diff = (s - o).abs().max()
    mean_diff = (s - o).abs().mean()
    print(f'\n  {name}')
    print(f'    Max  absolute diff : {max_diff:.4f}')
    print(f'    Mean absolute diff : {mean_diff:.4f}')

print('=' * 65)
print('DISTRIBUTION COMPARISON SUMMARY')
print('=' * 65)

compare_distributions(df['Primary Type'], df_sampled['Primary Type'], 'Primary Type')
compare_distributions(df['Description'], df_sampled['Description'], 'Description')
compare_distributions(df['Year'], df_sampled['Year'], 'Year')
compare_distributions(df['Location Description'], df_sampled['Location Description'], 'Location Description')
compare_distributions(df['Latitude'], df_sampled['Latitude'], 'Latitude')
compare_distributions(df['Longitude'], df_sampled['Longitude'], 'Longitude')

print(f'\n  Arrest rate')
print(f'    Original : {df["Arrest"].mean():.4f}')
print(f'    Sampled  : {df_sampled["Arrest"].mean():.4f}')
print(f'    Diff     : {abs(df["Arrest"].mean() - df_sampled["Arrest"].mean()):.4f}')
```

## Severity Mapping berdasarkan Primary Type

```python
# Pemetaan bobot ancaman berdasarkan Primary Type
severity_mapping = {
    'HOMICIDE': 100.0, 'ASSAULT': 70.0, 'BATTERY': 65.0,
    'ROBBERY': 80.0, 'BURGLARY': 60.0, 'THEFT': 40.0,
    'MOTOR VEHICLE THEFT': 50.0, 'NARCOTICS': 45.0, 'DECEPTIVE PRACTICE': 30.0
}
df_sampled['Severity_Score'] = df_sampled['Primary Type'].map(severity_mapping).fillna(25.0)
```

## Spatial Temporal & BallTree Indexing

```python
from sklearn.neighbors import BallTree
from tqdm.auto import tqdm

# Definisikan Parameter Space-Time Decay
LAMBDA_TIME = 0.05
GAMMA_SPACE = 1.0

# Tentukan Time Buckets Harian
start_date = df_sampled['Parsed_Date'].min().floor('D')
end_date = df_sampled['Parsed_Date'].max().floor('D')
target_dates = pd.date_range(start=start_date, end=end_date, freq='D')

# Spatial Clustering dari df_sampled
cluster_centers = df_sampled[['Latitude', 'Longitude']].drop_duplicates().sample(100, random_state=42).values

# Membangun BallTree
coords_rad = np.radians(df_sampled[['Latitude', 'Longitude']].values)
tree = BallTree(coords_rad, metric='haversine')
center_rad = np.radians(cluster_centers)
neighbors_indices = tree.query_radius(center_rad, r=2.0 / 6371.0)
```

# 4. Psuedo Labelling

Karena dataset mentah Chicago Crimes tidak menyediakan variabel *Risk Score* secara langsung, kami merancang sistem *pseudo-labeling* hibrid yang menggabungkan faktor kriminologi spasial-temporal dengan penguat kontekstual (*contextual amplifiers*).

Formula ini dirancang untuk mencerminkan tiga dimensi risiko utama secara kuantitatif: **Tingkat Keparahan (*Severity*)**, **Dimensi Waktu (*Temporal Decay*)**, dan **Dimensi Ruang (*Spatial Decay*)**.

### 1. Formula Matematis Utama

Perhitungan *Risk Score* mentah untuk suatu area spasial tertentu pada hari target dirumuskan sebagai berikut:

$$\text{Risk Score} = \left( \sum_{k} \text{Severity_Score}_k \times w_{t, k} \times w_{s, k} \right) \times \text{Weekend_Modifier} \times \text{Spatial_Density_Amplifier}$$

### 2. Penjelasan Komponen

* **$\text{Severity\_Score}_k$**: Bobot ancaman dasar dari insiden kejahatan historis ke-$k$, yang ditentukan berdasarkan jenis kejahatan (*Primary Type* / *FBI Code*) yang melibatkan ancaman fisik atau kerugian properti yang signifikan.
* **$w_{t, k}$ (Time Decay Weight):**

$$w_{t, k} = e^{-\lambda \Delta t}$$



Fungsi peluruhan eksponensial waktu di mana $\Delta t$ adalah selisih hari antara hari evaluasi dan waktu kejadian historis. Parameter laju peluruhan disetel $\lambda = 0.05$ untuk merefleksikan bahwa dampak insiden kejahatan masa lalu akan semakin memudar seiring berjalannya waktu.
* **$w_{s, k}$ (Spatial Decay Weight):**

$$w_{s, k} = e^{-\gamma \Delta d}$$



Fungsi peluruhan eksponensial jarak di mana $\Delta d$ adalah jarak fisik aktual dalam kilometer (dihitung menggunakan formula *Haversine* melalui struktur pencarian *BallTree* yang efisien). Parameter laju peluruhan disetel $\gamma = 1.0$.
* **$\text{Weekend\_Modifier}$**: Faktor pengali kontekstual senilai $1.2$ jika hari target jatuh pada akhir pekan (Sabtu/Minggu) dan $1.0$ pada hari biasa, untuk mengakomodasi peningkatan mobilitas dan potensi kerawanan sosial.
* **$\text{Spatial\_Density\_Amplifier}$**: Faktor penyesuaian dinamis berdasarkan volume riwayat kejahatan lokal pada area tersebut untuk memastikan wilayah dengan kluster padat mendapatkan kalibrasi risiko yang proporsional.

### 3. Normalisasi Target (*Min-Max Scaling*)

Setelah seluruh skor mentah terkumpul dari iterasi spasial-temporal, skor tersebut dinormalisasi ke dalam rentang baku **1.0 hingga 100.0** menggunakan formula *Min-Max Scaler*:

$$\text{Normalized\_Risk\_Score} = 1.0 + \left( \frac{\text{Risk\_Score} - \text{Score}_{\min}}{\text{Score}_{\max} - \text{Score}_{\min}} \right) \times 99.0$$

```python
pseudo_labels = []

for cluster_id in tqdm(range(len(cluster_centers)), desc="Processing Spatial Clusters"):
    center_lat, center_lon = cluster_centers[cluster_id]
    neighbor_idx = neighbors_indices[cluster_id]

    if len(neighbor_idx) == 0:
        for t_date in target_dates:
            pseudo_labels.append({'Date': t_date, 'Spatial_Cluster_ID': cluster_id, 'Risk_Score': 0.0})
        continue

    history_subset = df_sampled.iloc[neighbor_idx][['Parsed_Date', 'Latitude', 'Longitude', 'Severity_Score']].copy()

    # Haversine Distance Calculation
    lat1, lon1 = np.radians(center_lat), np.radians(center_lon)
    lat2, lon2 = np.radians(history_subset['Latitude'].values), np.radians(history_subset['Longitude'].values)

    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.sqrt(a))
    history_subset['Distance_km'] = 6371 * c

    for target_date in target_dates:
        past_events = history_subset[
            (history_subset['Parsed_Date'] < target_date) &
            (history_subset['Parsed_Date'] >= target_date - pd.Timedelta(days=30))
        ]

        if past_events.empty:
            risk_score = 0.0
        else:
            delta_t_days = (target_date - past_events['Parsed_Date']).dt.total_seconds() / (24 * 3600)
            w_t = np.exp(-LAMBDA_TIME * delta_t_days.values)
            w_s = np.exp(-GAMMA_SPACE * past_events['Distance_km'].values)
            risk_score = np.sum(past_events['Severity_Score'].values * w_t * w_s)

        pseudo_labels.append({
            'Date': target_date,
            'Spatial_Cluster_ID': cluster_id,
            'Risk_Score': risk_score
        })
```

## Normalization

```python
target_df = pd.DataFrame(pseudo_labels)
min_score = target_df['Risk_Score'].min()
max_score = target_df['Risk_Score'].max()

if max_score > min_score:
    target_df['Normalized_Risk_Score'] = 1.0 + ((target_df['Risk_Score'] - min_score) / (max_score - min_score)) * 99.0
else:
    target_df['Normalized_Risk_Score'] = 1.0

# Simpan ke CSV
target_df.to_csv('../chicago_risk_pseudo_labels.csv', index=False)
display(target_df.head())
```
